### 📌 Lámina 5: Definición del Problema (Físico vs. Machine Learning)

<img src="diapositivas_img/slide_05.png" width="680" alt="Diapositiva 5" />

- Para definir formalmente nuestra solución, pongamos sobre la mesa a nuestros dos grandes protagonistas con sus fortalezas y debilidades.
- Por un lado está nuestro modelo base NOCT, que nos entrega mucha transparencia y base termodinámica, pero que como vimos es muy rígido y asume condiciones ideales que no siempre se cumplen.
- Y por otro lado tenemos al algoritmo XGBoost, que es excelente encontrando patrones complejos en los datos, pero que por sí solo opera como una caja negra ciega a la física.
- La forma elegante de unir a estos dos mundos sin que se estorben es lo que en la literatura se llama "Aprendizaje Residual".
- Esto significa que no vamos a poner a la inteligencia artificial a predecir la potencia desde cero, porque para dar un piso seguro ya tenemos las ecuaciones físicas.
- Lo que haremos será usar el modelo NOCT para hacer el trabajo pesado y estimar la energía base de forma interpretable.
- Y luego dejaremos que XGBoost entre a la cancha exclusivamente para predecir el residuo, es decir, aprender únicamente de los errores que comete el modelo físico cuando la luz se dispersa.
- Al final, la suma de la potencia calculada por la física más esta corrección inteligente del residuo, es lo que nos da la estimación de nuestra Solución Híbrida.
- Esta combinación nos garantiza lo mejor de ambos mundos: el resguardo de las leyes termodinámicas y la altísima precisión analítica de los datos puros.
- PIE: Y para lograr articular todo este modelo en la práctica, tuvimos que trazar una ruta metodológica clara, que es lo que veremos en los objetivos del trabajo...

### 📌 Lámina 7: Fundamento Físico: Fracción Difusa (fd)

<img src="diapositivas_img/slide_07.png" width="680" alt="Diapositiva 7" />

- ya metiendonos de lleno en el marco teorico hay un concepto que es el corazon de esta tesis y es la irradiancia y como se compone en la realidad
- la radiacion global que llega al panel que nosotros llamamos ghi no es un rayo unico de luz sino que esta formada por dos componentes principales que actuan super distinto
- por un lado tenemos la radiacion directa que es ese rayo de sol puro que te pega directo en la cara y que te genera una sombra super marcada en el suelo
- y por otro lado esta la radiacion difusa que es toda esa luz que choca y se dispersa por las nubes o las particulas en la atmosfera y te llega rebotando de todos lados a la vez
- aca es donde entra la estrella del trabajo que es la fraccion difusa que la calculamos super facil simplemente dividiendo la radiacion difusa por la global total
- este simple numerito es como un termometro optico del cielo que nos dice al instante que tan nublado o despejado esta el dia en ese momento exacto
- si la fraccion difusa nos da cero significa que el cielo esta completamente despejado sin una nube y toda la luz te llega de forma directa y limpia
- en cambio si el valor se acerca a uno significa que el dia esta totalmente cubierto o nublado y practicamente toda la luz que llega esta dispersa y rebotando
- lo genial de usar este valor de fraccion difusa es que nos permite clasificar el comportamiento de la luz matematicamente sin tener que adivinar mirando al cielo porque solo viendo este numero ya sabemos a que condicion optica nos enfrentamos

### 📌 Lámina 8: Regímenes Radiativos y Filtros Físicos

<img src="diapositivas_img/slide_08.png" width="680" alt="Diapositiva 8" />

- teniendo claro el concepto de la fraccion difusa lo que hicimos fue usar ese numero para separar nuestro año de datos meteorologicos en tres escenarios o regimenes distintos
- el primer escenario es cuando tenemos una fraccion difusa baja o sea menor a 0.3 que basicamente son esos dias de pleno sol casi sin nubes donde manda la radiacion directa
- despues definimos un regimen mixto que va entre 0.3 y 0.7 que son estos dias medios inestables o parcialmente nublados que le complican muchisimo la vida a los pronosticos normales
- y por ultimo tenemos el regimen de alta fraccion difusa que es mayor a 0.7 y esos son derechamente los dias de cielo totalmente cubierto o dias super grises de invierno
- pero antes de poner a los modelos a trabajar a lo loco tuvimos que aplicar un filtro fisico super estricto a los datos porque no todas las horas del dia nos servian para evaluar el modelo
- imaginen que al amanecer o al atardecer hay muy poquita radiacion solar y esos valores tan chicos vuelven locas a las divisiones matematicas y te generan puros ruidos falsos en los graficos
- asi que cortamos por lo sano y decidimos dejar solamente las horas donde la radiacion global y la que le incide al panel superaran los 10 watts por metro cuadrado que es como el umbral donde recien empieza la generacion
- asi que de las 8760 horas totales que tiene un año nos terminamos quedando con exactamente 4215 horas de sol efectivo diurno puro donde realmente hay generacion importante en santiago
- con esta limpieza tremenda de los datos nos aseguramos de que el modelo hibrido despues aprenda solo de datos fisicamente reales y utiles y no de puros errores de calculo que pasan de noche o en la penumbra

### 📌 Lámina 9: Modelo Físico Base (NOCT) y Limitaciones

<img src="diapositivas_img/slide_09.png" width="680" alt="Diapositiva 9" />

- ahora pasamos a hablar del famoso modelo fisico base que usamos aca que es el estandar de la industria el super clasico modelo noct
- este modelo es el que todos los softwares usan porque es super interpretable y facil de entender pero tiene unas limitantes fisicas super fuertes que son justo las que nosotros queremos atacar en esta tesis
- por ejemplo el noct asume ciertas cosas por defecto que no siempre pasan en la vida real como que asume que siempre hay un viento constante de 1 metro por segundo y deja totalmente de lado el enfriamiento real que produce el viento pegandole al panel
- y la otra limitacion que es critica para nosotros es su limitacion optica porque este modelo basico trata a toda la luz solar exactamente igual y le da lo mismo si viene directa o si viene rebotando de una nube
- fisicamente eso es un error grave porque sabemos que la radiacion difusa incide desde todos los angulos posibles y al golpear asi el vidrio del panel hace que rebote muchisima mas luz y se pierda mas energia util
- pero el modelo noct como no sabe separar estos tipos de luz simplemente asume que todo entra perfecto al panel y eso termina generando errores feos de sobreestimacion sobre todo en los dias nublados
- aca es super importante mencionar que usamos los parametros por defecto de pvwatts como la temperatura nominal de 45 grados y el 14 por ciento de perdidas globales para tener un punto de referencia totalmente estandar
- nuestra meta aca no era armar el modelo fisico mas perfecto del mundo sino tomar el que usa normalmente la industria y demostrar matematicamente donde cojea y como la inteligencia artificial puede entrar a tapar esos huecos que la fisica clasica asume de una forma muy basica
